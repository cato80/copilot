##### Supraconductivité et supraluminique

Plus la résistance du vide est faible, plus la vitesse de la lumière est grande : La célérité est définie pour une résistance (impédance) du vide de $Z=376 \Omega$, j'ai établi une équation qui permet de calculer une vitesse imaginaire supraluminique, que j'appelle chiracécité $ç=314159265$ d'une résistance de $360 \Omega$.
