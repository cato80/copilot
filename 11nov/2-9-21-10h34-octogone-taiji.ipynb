###### Définition du taiji

C'est la racine carrée de l'accélération(négative) moins la dérivée :

Accélération : $x^2$ ou $c^2$
Vitesse : $2x$ ou $2c$

```python
def f(x):
    if x<0:
        return sqrt(-x^2-2*x)
    else:
        return -sqrt(-x^2+2*x)
a=plot(f,(x,-2,2),color='white',fill=sqrt(4-x^2),
fillcolor="lightgray",alpha=0.1)
p=plot(f,(x,-2,2),color='white',fill=-sqrt(4-x^2),
fillcolor="lightcyan",alpha=0.1)
b = circle((0,0),2,color="salmon",alpha=0.1)
show(a+p,figsize=2,axes=False,aspect_ratio=1)
```
