##### Version électronique de la circonférence

La circonférence du cercle $c=\pi r2$ est égale à $\omega=\pi f2$ ou $w=2\pi\nu$.

```python
%display latex
r=2;c=pi*r;c
```

Ainsi la longueur d'onde fondamentale est $2\pi=6.28$

Les scientifiques ont assimilé la célérité $c$ à la circonférence $\lambda$.

```python
show(LatexExpr(r"\lambda=\frac{\frac{1}{\sqrt\mu\epsilon}}{\frac{1}{2\pi\sqrt\mu\epsilon}} = \frac{c}{\nu}=2\p}{\frac{1}{2\pi\sqrt\mu\epsilon}} = \frac{c}{\nu}=2\pi"))
```

La vraie valeur non polarisé de $c=\frac{1}{\sqrt\mu\epsilon}=314159265=ç$ 

```python
L=1/n(pi)*360E-8;C=1/(n(pi)*360E+8);
ç=1/sqrt(L*C);round(ç,8)
```

```python
show(LatexExpr(r"ç=\frac{1}{\sqrt L C }=314159265"))
```

Valeur scientifique de $c$, avec la coudée du pied du roi

```python
rr=2/sqrt(n(pi));f(x)=sqrt(rr^2-x^2);base=439.995154;round(base,6)
```

```python
# valeur cercle circonscrit
c=n(pi)/6;r=sqrt(2.0);cp=c*r*n(pi)*439.995154;cp
```

```python
# cercle inscrit
cr=n(pi)/6;rp=c*n(pi)*439.995154;rp
```

```python
# coudée du pied du roi
cc=439.995154;ll=cr*cc*r*n(pi);ll
```

```python
# vc
lc=cr*cc*n(pi);round((ll-lc)*1000000,8)
```
