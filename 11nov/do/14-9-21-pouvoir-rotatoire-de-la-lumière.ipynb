###### L'espace n'est pas un milieu homogène, la célérité change

L'espace (éther) est l'impédance du vide qui est différente de zéro, une impédance zéro est un trou noir, c'est-à-dire une résonance inductive/capacitive.

La lumière est un microscope/téléscope qui grossit les atomes en planètes.

Les couleurs proviennent du pouvoir rotatoire de la lumière
