```python
def f(x):
    if x<0:
        return sqrt(-x^2-2*x)
    else:
        return -sqrt(-x^2+2*x)
a=plot(f,(x,-2,2),color='white',fill=sqrt(4-x^2),
fillcolor="lightgray",alpha=0.1)
b=plot(f,(x,-2,2),color='white',fill=-sqrt(4-x^2),
fillcolor="lightyellow",alpha=0.1)
p = circle((0,0),2,color="salmon",alpha=0.1)
show(a+b+p,figsize=3,axes=False,aspect_ratio=1)
```
