###### Analogie célérité $c$ et RLC

L'équation de la célérité d'une particule a caché une équation en arrière plan, plus importante : RLC. Le vide, autrefois appelé éther, est une impédance (résistance) au mouvement, déplacement, vitesse. On peut imaginer l'espace comme un circuit RLC : R égale le vide (en ohms, impédance dans le cas de l'alternatif), L inductance égale la perméabilité du vide ($\mu$) et C capacitance égale la permittivité du vide ($\epsilon$).

```python
show(LatexExpr(r"c=\frac{1}{\sqrt \mu \epsilon}=299792458"))
show(LatexExpr(r"ç=\frac{1}{\sqrt \mu \epsilon}=314159265"))
```

```python
show(LatexExpr(r"ç=\frac{1}{\sqrt L C}=314159265"))
show(LatexExpr(r"c=\frac{1}{\sqrt L C}=299792458"))
```

```python
c=299792458;mu=n(pi)*4E-7;show(LatexExpr(r"L:\mu="),mu)
epsilon=1/(mu*c*c);show(LatexExpr(r"C:\epsilon="),epsilon)
```

```python
L=1/n(pi)*360E-8;show(LatexExpr(r"L:\mu="),L)
C=1/(n(pi)*360E+8);show(LatexExpr(r"C:\epsilon="),C)
```

```python
rp=mu*epsilon*L*C
```

```python
pr=2/sqrt(n(pi));r*1E-34
```
