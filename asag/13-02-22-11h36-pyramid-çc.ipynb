##### géométrie et physique classique

La différence entre un cercle circonscrit et un cercle inscrit, dont la base est de 440 Hz (exactement 439.995154) donne la constante de la fréquence de la lumière à l'échelle du millionième. La lumière est bien un rapport de fréquence avec une fondamentale (la fréquence de résonance).  

```python
%display latex
cr=n(pi)/6;r=sqrt(2.0) # coudée royale
cc=439.995154 #base 440
lg=cr*r*n(pi)*439.995154;lg # cercle circonscrit en MHz
pg=cr*cc*n(pi);lg,pg
```

```python
c=lg-pg;int(c*1000000.000000)# mètres
```

```python
# micro mètres (Tera herz THz)
c=lg-pg;int(c*1000000000000)
```

```python
c=lg-pg;int(c*1000000000000000)# nano mètres
```

```python
# nano mètres Peta Hertz PHz
c=lg-pg;int(c*1000000000000000)# nano mètres
```

$L=\large \mu$ et $C=\large \epsilon$, inductance et capacitance, équivalence de la perméabilité du vide et de la permittivité.

```python
a=360E-8;b=360E+8;L=1/n(pi)*a;C=1/(n(pi)*b);ç=1/sqrt(L*C)
show(LatexExpr(r"\large \mu=L=\frac{1}{\pi a}="),L)
show(LatexExpr(r"\large \epsilon=C=\frac{1}{\pi b}="),C)
show(LatexExpr(r"\large ç=\frac{1}{\sqrt L C}="),ç)
show(LatexExpr(r"\large c=\frac{1}{\sqrt \mu \epsilon}="),c)
```
