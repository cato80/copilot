```python
# Importation de la fonction tangente
from math import tan, radians

# Angle en degrés
theta_c_degrees = 42

# Conversion en radians
theta_c_radians = radians(theta_c_degrees)

# Calcul des coordonnées x et y
x = 1
y = x * tan(theta_c_radians)

# Création du graphique
p = points([(0,0), (x,y)], size=30, color='blue', axes_labels=['$x$', '$y$'])

# Tracer la ligne rouge pointillée
l = line([(0,0), (x,y)], linestyle='dashed', color='red', legend_label='oc=42°')

# Afficher le graphique
show(p + l)
```
