```python
# Constantes utilisées dans l'expérience
Qe = 299792458  # Charge de la bouteille de Leyde en unités électrostatiques
Qm=1  # Charge mesurée en unités électromagnétiques
rapport_Qe_Qm = Qe / Qm  # Rapport Qe/Qm

# Conversion du rapport en vitesse
c = 299792458  # Vitesse de la lumière en m/s (valeur actuelle, pas celle utilisée à l'époque)
vitesse_estimee = rapport_Qe_Qm * c

# Affichage des résultats
print(f"Rapport Qe/Qm : {rapport_Qe_Qm}")
print(f"Vitesse estimée : {vitesse_estimee} m/s")
```
